**Loading necessary libraries**

```python
import numpy as np
import requests 
from io import BytesIO
from pathlib import Path
```

**Fetching Data**

```python
cat_url="https://storage.googleapis.com/quickdraw_dataset/full/numpy_bitmap/cat.npy"
clock_url="https://storage.googleapis.com/quickdraw_dataset/full/numpy_bitmap/clock.npy"

cats = np.load(BytesIO(requests.get(cat_url).content))
clocks = np.load(BytesIO(requests.get(clock_url).content))

print(cats.shape)
print(clocks.shape)
```

**Saving data**

```python
raw_dir = Path("../data/raw")
raw_dir.mkdir(parents=True, exist_ok=True)
np.save(raw_dir/"cat.npy", cats)
np.save(raw_dir/"clock.npy", clocks)
```
